```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
#Input sentence
sentence = ["I", "Love", "Deep", "Learning"]
print(sentence)
```

```python
#word Embeddings
word_embeddings = {
    "I": 1.0,
    "Love": 2.0,
    "Deep": 3.0,
    "Learning": 4.0
}
print(word_embeddings)
```

```python
x = np.array([
    [
    [word_embeddings["I"]],
    [word_embeddings["Love"]],
    [word_embeddings["Deep"]]
    ]
])
print(x)
```

```python
y = np.array([
    [word_embeddings["Learning"]]
])
print(y)
```

```python
print("Input shape: ", x.shape)
print("Output shape: ", y.shape)
```

```python
model = Sequential()

model.add(
LSTM(
units=8,
activation='tanh',
input_shape=(3,1)
)
)
model.add(Dense(1))

model.summary()
```

```python
#compile model
model.compile(
    optimizer="adam",
    loss='mse'
)
```

```python
#Train the model
model.fit(x, y, epochs=300, verbose=1)
```

```python
#Prediction
test_sequence = np.array([
    [
        [1.0],
        [2.0],
        [3.0]
    ]
])

prediction=model.predict(test_sequence)
print(prediction)
```

```python
#Convert Prediction to Word

predicted_number=prediction[0][0]
closest_word=min(
word_embeddings,
key=lambda word:abs(word_embeddings[word]-predicted_number)
)

print("Predicted Word:", closest_word)
```
