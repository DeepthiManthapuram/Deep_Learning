# Downloading dataset

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("lorenzoarcioni/road-damage-dataset-potholes-cracks-and-manholes")

print("Path to dataset files:", path)
```

# Import libraries

```python
import os
import cv2
import pandas as pd
import matplotlib.pyplot as plt
```

# Check dataset folder

```python
print(os.listdir(path))
```

```python
data_path = os.path.join(path, "data")

print(os.listdir(data_path))
```

# Displaying first 5 images

```python
images_path = os.path.join(data_path, "images")

print(os.listdir(images_path)[:5])
```

# Count total number of images

```python
image_files = []

for file in os.listdir(images_path):

    if file.endswith((".jpg", ".png", ".jpeg")):

        image_files.append(file)

print("Total Images:", len(image_files))
```

# Identify number of classes

```python
import json

# Path to the COCO annotation file
annotations_file_path = os.path.join(data_path, "annotations_coco.json")

classes = []

if os.path.exists(annotations_file_path):
    with open(annotations_file_path, 'r') as f:
        coco_data = json.load(f)

    # Extract category names (classes) from the 'categories' list
    if 'categories' in coco_data:
        classes = [category['name'] for category in coco_data['categories']]
    else:
        print("No 'categories' key found in the annotation file.")
else:
    print(f"Annotation file not found: {annotations_file_path}")

print("Classes:", classes)
print("Number of Classes:", len(classes))
```
