# **Tasks**

```python
import nltk
import re
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

# Task1: text cleaning

```python
text = "Hello!!! I am learning NLP @ CVR College :)"

# Lowercase
text = text.lower()

# Remove punctuation
text = re.sub(r'[^\w\s]', '', text)

# Tokenize
words = word_tokenize(text)

# Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

# Join words
cleaned_text = " ".join(filtered_words)

print(cleaned_text)
```

# Task2:tokenizqation Analyzer

```python
paragraph = """
Natural Language Processing is an important field.
It helps computers understand human language.
"""

# Sentence Tokenization
sentences = sent_tokenize(paragraph)

# Word Tokenization
words = word_tokenize(paragraph)

print("Sentences:")
print(sentences)

print()

print("Words:")
print(words)

print()

print("Total Sentences:", len(sentences))

print("Total Words:", len(words))

print("Unique Words:", len(set(words)))
```

# Task3:Stopword Removal Tool

```python
text = "This is an amazing NLP session for the students"

# Tokenize
words = word_tokenize(text)

# Stopwords
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in words:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print(" ".join(filtered_words))
```

# Task4:Stemming vs Lemmitization

```python
words = ["running", "studies", "caring"]

# Stemming
stemmer = PorterStemmer()

stemmed_words = []

for word in words:
    stemmed_words.append(
        stemmer.stem(word)
    )

print("Stemmed Words:")
print(stemmed_words)

print()

# Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = []

for word in words:
    lemmatized_words.append(
        lemmatizer.lemmatize(word)
    )

print("Lemmatized Words:")
print(lemmatized_words)
```

# Task5:Bag of words Implementation

```python
documents = [
    "I love NLP",
    "NLP is interesting",
    "I love machine learning"
]

# Create Vectorizer
vectorizer = CountVectorizer()

# Fit and Transform
X = vectorizer.fit_transform(documents)

# Vocabulary
print("Vocabulary:")
print(vectorizer.vocabulary_)

print()

# Document Vectors
print("Document Vectors:")

print(X.toarray())
```

# Task6:TF-IDF Important word Detection

```python
documents = [
    "NLP is useful in search engines",
    "Search engines use NLP",
    "TF IDF helps identify important words"
]



# TF-IDF Vectorizer
tfidf = TfidfVectorizer()

X = tfidf.fit_transform(documents)

# Feature Names
features = tfidf.get_feature_names_out()

print("Important Words:")

print(features)

print()

print("TF-IDF Matrix:")

print(X.toarray())
```
